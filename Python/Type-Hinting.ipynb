## Differnece between Type checking, Data validation and Type hinting

## 1. Type Hinting
Definition: Type hinting is a way of specifying the expected data types of variables, function parameters, and return values in Python.

It helps developers understand what type of data a function expects. 

Here:
- a: int → Expected input is an integer.
- b: int → Expected input is an integer.
- -> int → Expected return type is an integer.
Important: Python does not enforce type hints at runtime by default.


```python
def add_numbers(a: int, b: int) -> int:
    return a + b

result = add_numbers(10, 20)
print(result)
```

## 2.Type Checking
Definition: Type checking is the process of checking whether variables, function arguments, and return values match their expected data types.

It helps identify type-related errors in code.

```python
def add_numbers(a: int, b: int) -> int:
    return a + b

result = add_numbers("10", 20)
```

A static type checker such as Mypy can identify that "10" is a string instead of an integer.

Common tools:
- Mypy
- Pyright

## 3. Data Validation
Definition: Data validation is the process of checking whether incoming data meets predefined rules, constraints, and requirements before processing it.

It ensures that the data is valid and acceptable for your application.

# TYPE HINTING 
Type hints are annotations used to indicate the expected data types of variables, function parameters, and return values. They improve code readability and help developers understand how code should be used.

### 1. Optional Types 
Optional is a type hint that indicates a variable or function parameter can contain either a specified data type or None

```python
def get_username(name: str | None) -> str:
    if name is None :
        return "no value "
    return name
print(get_username("yashhh"))
print(get_username(None))
```

### 2. Tuple Type Hints
A tuple is an ordered, immutable collection of elements. Tuple type hints specify the expected types and structure of those elements.

```python
def details(name : str , age:int , marks :float) -> tuple[str, int, float]:
    return name , age ,marks 
result =  details("Yasaswi", 20, 85.5)
print(result)
```

```python
from typing import NewType 
```

## What is TypedDict?
TypedDict is a feature in Python's typing module that allows you to define the expected structure of a dictionary by specifying the types of its keys and corresponding values.
Which keys should be present.
What type of value each key should contain.

```python
from typing import TypedDict

class Student(TypedDict):
    name: str
    age: int
    marks: float


student: Student = {
    "name": "Yasaswi",
    "age": 20,
    "marks": 92.5
}
print(student)
```

## Generic Types
Generics allow you to write reusable functions, classes, and data structures that work with different data types while maintaining type safety.
##### Python provides TypeVar through the typing module.

```python
from typing import TypeVar

T = TypeVar("T")

def get_first(items: list[T]) -> T:
    return items[0]
numbers = [10, 20, 30]
names = ["Yasaswi", "khaleel", "Kiran"]

print(get_first(numbers))  # 10
print(get_first(names))    # Yasaswi
```
